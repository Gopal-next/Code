Sabse pehle, maine `Node` naam ki class banayi. Class ek blueprint hoti hai jisse hum objects banate hain.

Uske andar `__init__()` method hai, jo naye node ki initial values set karta hai.

`self.data = data` node ki value store karta hai, aur `self.next = None` ka matlab hai ki shuru mein node kisi doosre node se connected nahi hai.

Phir maine teen objects banaye: `first`, `second`, aur `third`, jinmein 10, 20 aur 30 store hain.

Uske baad, `first.next = second` se first node ko second node se connect kiya, aur `second.next = third` se second node ko third node se connect kiya.

Finally, maine `print()` ki help se har node ka data access kiya. Last node ka `next` None hai, kyunki uske baad koi node nahi hai.

```python
class Node:
    def __init__(self, data):
        self.data = data
        self.next = None

first = Node(10)
second = Node(20)
third = Node(30)

first.next = second
second.next = third

print(first.next)
print(first.next.data)
print(first.next.next.data)
print(third.next)
```

### Traversing a Linked List

```python
class Node:
    def __init__(self, data):
        self.data = data
        self.next = None

first = Node(10)
second = Node(20)
third = Node(30)

first.next = second
second.next = third

current = first
while current is not None:
    print(current.data, end = ' -> ')
    current = current.next
print("null")
```

#### Create a Student class
##### Easy
Create a Student class with name and age.
- Store both values using __init__().
- Create two objects with different names and ages.
- Print the details of both student

Maine `Student` naam ki ek class banayi hai, jo ek blueprint hai. Uske andar `__init__()` ek special method hai jo object create hone par automatically call hota hai aur initial values set karta hai. Isme `self` current object ko refer karta hai, aur `name` aur `age` parameters hain jinmein values receive hoti hain. `self.name` aur `self.age` attributes hain jo har object ka data separately store karte hain. Jab maine `s1 = Student("Gopal", 23)` likha, toh ek object create hua aur `"Gopal"` aur `23` arguments ke roop mein pass hue. Isi tarah `s2` ek alag object hai jisme Aman ka data hai. Finally, `s1.name` aur `s1.age` jaise dot notation ka use karke maine object ke attributes access kiye aur print kar diye.

```python
class Student:
    def __init__(self, name, age):
        self.name= name
        self.age = age

s1 = Student("Gopal", 23)
s2 = Student("Aman",45)

print(s1.name, s1.age)
print(s2.name , s2.age)
```

### Create a Rectangle class
###### Easy
Create a Rectangle class with length and width.
- Store both values in the constructor.
- Create a method area() that returns length × width.
- Create an object with length 10 and width 5.
- Print its area.

```python
class Rectangle:
    def __init__(self, l, b):
        self.l = l
        self.b = b

    def area(self):
        return self.b * self.l

r1 = Rectangle(3,4)

print(r1.area())

    
```

### Create a BankAccount class
#### Easy–Medium
Create a BankAccount class with owner and balance.
- Initialize the account with an owner's name and starting balance.
- Create a deposit(amount) method that adds money to the balance.
- Create a show_balance() method that prints the current balance.
- Deposit 500 into an account starting with 1,000, then display the balance.

```python
class BankAccount:
    def __init__(self, name, balance):
        self.name = name
        self.balance = balance

    def deposit(self, amount):
        self.balance += amount

    def show_balance(self):
        return self.balance

s1 = BankAccount("Gopal", 1000)
s1.deposit(500) 
print(s1.show_balance())
```
